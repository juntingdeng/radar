```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation

# Saved from: np.asarray(all_bevs)
# Shape: (n_frames, H, W), each frame is already bev_cartesian.T
all_bevs = np.load("../processed/test/all_bev_180s.npy")

# # Same xlin, ylin used in your PyQtGraph setRect()
# xlin = np.load("xlin.npy")
# ylin = np.load("ylin.npy")

fps = 10
out_file = "../processed/test/bev_video_180s.mp4"

vis = np.asarray(all_bevs).transpose(0, 2, 1)

# xmin = xlin.min() * 0.5
# xmax = xlin.max() * 0.5
# ymin = ylin.min() * 0.5
# ymax = ylin.max() * 0.5

# Fixed display range to avoid flickering
vmin = np.percentile(vis, 1)
vmax = np.percentile(vis, 99)

fig, ax = plt.subplots(figsize=(7, 7))

im = ax.imshow(
    vis[0],
    origin="lower",
    # # extent=[xmin, xmax, ymin, ymax],
    # aspect="equal",
    # vmin=vmin,
    # vmax=vmax,
)

ax.set_xlabel("x [m]")
ax.set_ylabel("y [m]")
title = ax.set_title("Frame 0")
# plt.colorbar(im, ax=ax, label="BEV intensity")

def update(i):
    im.set_data(vis[i])
    title.set_text(f"Frame {i}")
    return im, title

ani = animation.FuncAnimation(
    fig,
    update,
    frames=vis.shape[0],
    interval=1000 / fps,
    blit=False,
)

ani.save(out_file, writer="ffmpeg", fps=fps, dpi=150)
plt.close(fig)

print(f"Saved video to {out_file}")
```
