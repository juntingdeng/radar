```python
idle = 10 #us 
ramp = 120 #us 
nchirps = 3 #us 
nloops = 16 #us 
frame = 100*1e3 #us 
nsamples = 1024
samplerate = 10000 # ksps
slope = 29.982 #GHz/us
c = 3*1e8 #m/s

framerate = 1/(frame*1e-6)
active = (idle+ramp) * nchirps * nloops
dutycycle = active / frame

# bandwidth = slope * (nsamples/samplerate) <- slope * Adc_active_time
# range resolution = c / (2*bandwidth)
# range maximum = nsamples * resolution
#               = nsamples * c / (2*bandwidth) 
#               = c * samplerate / (s*slope)
bandwidth = (slope*1e12) * (nsamples/(samplerate*1e3)) / 1e9 # GHz
rng_res = c/(2*(bandwidth*1e9)) # m
rng_max = c*(samplerate*1e3)/(2*(slope*1e12)) # m

print(f'frame rate: {framerate:.2f}f Hz \
       \nperiod: {frame*1e-3:.2f}ms \
        \nactive time: {active:.2f} us \
        \nduty cycle: {dutycycle*100:.2f}% \
        \nbandwidth: {bandwidth:.2f} GHz \
        \nrange resolution: {rng_res:.2f} m \
        \nrange max: {rng_max:.2f} m' 
        )
```
